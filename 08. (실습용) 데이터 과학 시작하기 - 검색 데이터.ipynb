### csv 파일을 파이썬으로 읽어오기



1.   파일 open
2.   파일 읽기 reader
3.   파일 헤더 건너뛰기
4.   파일을 리스트로 바꾸기



```
# csv 라이브러리 임포트
import csv

f = open('파일명.csv')         # 파일 오픈
data = csv.reader(f)         # 파일 읽기 reader
next(data)                   # 파일 헤더 건너뛰기
data = list(data)            # 파일 리스트로 바꾸기
```
